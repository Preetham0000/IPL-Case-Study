## Task 0: Data Loading and Preprocessing

```python
import numpy as np

# File paths
matches_path = r"..\dataset\matches.csv"
deliveries_path = r"..\dataset\deliveries.csv"

# numpy load gen txt
matches_np = np.genfromtxt(matches_path,delimiter=",",dtype=str,usecols=range(0,20))
print(matches_np[:5])

#numpy load gen txt 
deliveries_np = np.genfromtxt(deliveries_path,delimiter=",",dtype=str)
print(deliveries_np[:5])
```

## Task 1: Total Runs per Match

```python
# Task 1 uses deliveries_np loaded in Task 0
match_id = deliveries_np[1:, 0].astype(np.int64)
batsman_runs = deliveries_np[1:, 9].astype(np.int64)

unique_matches = np.unique(match_id)

total_runs_per_match = np.empty(
    len(unique_matches),
    dtype=[('match_id', np.int64), ('total_runs', np.int64)]
)

for i, mid in enumerate(unique_matches):
    total_runs_per_match[i] = (mid, np.sum(batsman_runs[match_id == mid]))

print("Match ID | Total Runs")
print("-" * 25)
for mid, total in total_runs_per_match[:10]:
    print(f"{mid:8d} | {total:10d}")
print(f"\n... ({len(total_runs_per_match)} total matches)")
```

## Task 2: Top 5 Batters

```python
batter = deliveries_np[1:, 6]
batsman_runs = deliveries_np[1:, 9].astype(np.int64)

unique_batters = np.unique(batter)

total_runs_per_batter = np.empty(
    len(unique_batters),
    dtype=[('batter', object), ('total_runs', np.int64)]
)

for i, player in enumerate(unique_batters):
    total_runs_per_batter[i] = (player, np.sum(batsman_runs[batter == player]))

sorted_batters = total_runs_per_batter[np.argsort(-total_runs_per_batter['total_runs'])]

print("Top 5 Batters by Total Runs")
print("-" * 40)
print(f"{'Player':<20} | {'Total Runs':>10}")
print("-" * 40)
for i in range(5):
    player, runs = sorted_batters[i]
    print(f"{player:<20} | {runs:>10d}")
```

## TASK 3: Strike Rate Calculation

```python
batter = deliveries_np[1:, 6]
batsman_runs = deliveries_np[1:, 9].astype(np.int64)

unique_batters = np.unique(batter)

strike_rate_data = np.empty(
    len(unique_batters),
    dtype=[('batter', object), ('strike_rate', np.float64)]
)

for i, player in enumerate(unique_batters):
    mask = batter == player
    total_runs = np.sum(batsman_runs[mask])
    balls = np.sum(mask)
    strike_rate_data[i] = (player, (total_runs / balls) * 100 if balls > 0 else 0)

sorted_data = strike_rate_data[np.argsort(-strike_rate_data['strike_rate'])]

print("Top 10 Batters by Strike Rate")
print("-" * 35)
print(f"{'Batter':<20} | {'Strike Rate':>10}")
print("-" * 35)
for i in range(10):
    player, sr = sorted_data[i]
    print(f"{player:<20} | {sr:>9.2f}%")
```

## TASK 4: Economy Rate of Bowlers


```python
bowler = deliveries_np[1:, 7]
total_runs = deliveries_np[1:, 11].astype(np.int64)

unique_bowlers = np.unique(bowler)

economy_data = np.empty(
    len(unique_bowlers),
    dtype=[('bowler', object), ('economy', np.float64)]
)

for i, player in enumerate(unique_bowlers):
    mask = bowler == player
    runs = np.sum(total_runs[mask])
    balls = np.sum(mask)
    overs = balls / 6
    economy_data[i] = (player, (runs / overs) if overs > 0 else 0)

sorted_data = economy_data[np.argsort(economy_data['economy'])]

print("Top 10 Bowlers by Economy Rate")
print("-" * 35)
print(f"{'Bowler':<20} | {'Economy':>10}")
print("-" * 35)
for i in range(10):
    player, econ = sorted_data[i]
    print(f"{player:<20} | {econ:>10.2f}")
```

## Task 5: Runs per Over (1 to 20)

```python
over = deliveries_np[1:, 4].astype(np.int64)
batsman_runs = deliveries_np[1:, 9].astype(np.int64)

avg_runs_per_over = np.zeros(20)

for i in range(1, 21):
    mask = over == i
    if np.sum(mask) > 0:
        avg_runs_per_over[i - 1] = np.mean(batsman_runs[mask])

print("Average Runs per Over")
print("-" * 30)
print(f"{'Over':<8} | {'Avg Runs':>10}")
print("-" * 30)
for over_num in range(1, 21):
    print(f"{over_num:<8} | {avg_runs_per_over[over_num - 1]:>10.2f}")
```

## Task 6: Boundary Analysis

```python
batsman_runs = deliveries_np[1:, 9].astype(np.int64)
batting_team = deliveries_np[1:, 2]

fours = np.sum(batsman_runs == 4)
sixes = np.sum(batsman_runs == 6)

print("Boundary Analysis")
print("-" * 30)
print(f"Total Fours: {fours}")
print(f"Total Sixes: {sixes}")
print(f"Total Boundaries: {fours + sixes}")

unique_teams = np.unique(batting_team)
team_boundaries = np.zeros(len(unique_teams))

for i, team in enumerate(unique_teams):
    team_runs = batsman_runs[batting_team == team]
    team_boundaries[i] = np.sum((team_runs == 4) | (team_runs == 6))

max_idx = np.argmax(team_boundaries)
print(f"\nTeam with most boundaries: {unique_teams[max_idx]} ({int(team_boundaries[max_idx])} boundaries)")
```

## Task 7: Death Overs Analysis

```python
over = deliveries_np[1:, 4].astype(np.int64)
total_runs = deliveries_np[1:, 11].astype(np.int64)
batting_team = deliveries_np[1:, 2]

death_mask = (over >= 16) & (over <= 20)

death_runs_total = np.sum(total_runs[death_mask])

print("Death Overs Analysis (Overs 16-20)")
print("-" * 40)
print(f"Total runs in death overs: {death_runs_total}")

unique_teams = np.unique(batting_team[death_mask])
team_death_runs = np.zeros(len(unique_teams))

for i, team in enumerate(unique_teams):
    team_death_runs[i] = np.sum(total_runs[(batting_team == team) & death_mask])

max_idx = np.argmax(team_death_runs)
print(f"Team with highest runs in death overs: {unique_teams[max_idx]} ({int(team_death_runs[max_idx])} runs)")
```

## Task 8: Highest Scoring Match

```python
match_id = deliveries_np[1:, 0].astype(np.int64)
total_runs = deliveries_np[1:, 11].astype(np.int64)

unique_matches = np.unique(match_id)
match_runs = np.zeros(len(unique_matches))

for i, mid in enumerate(unique_matches):
    match_runs[i] = np.sum(total_runs[match_id == mid])

max_idx = np.argmax(match_runs)

print("Highest Scoring Match")
print("-" * 40)
print(f"Match ID: {unique_matches[max_idx]}")
print(f"Total Runs: {int(match_runs[max_idx])}")
```

## Task 9: Match Winner Approximation

```python
match_id = deliveries_np[1:, 0].astype(np.int64)
batting_team = deliveries_np[1:, 2]
batsman_runs = deliveries_np[1:, 9].astype(np.int64)

unique_matches = np.unique(match_id)

print("Match Winner Approximation (Based on runs)")
print("-" * 50)

for mid in unique_matches[:5]:
    match_mask = match_id == mid
    teams = np.unique(batting_team[match_mask])

    print(f"\nMatch {mid}:")

    for team in teams:
        runs = np.sum(batsman_runs[(match_id == mid) & (batting_team == team)])
        print(f"  {team}: {runs} runs")
```

## Task 10: Toss Impact Analysis

```python
match_ids_matches = matches_np[1:, 0].astype(np.int64)
toss_winners = matches_np[1:, 9]

match_ids_del = deliveries_np[1:, 0].astype(np.int64)
batting_team = deliveries_np[1:, 2]
batsman_runs = deliveries_np[1:, 9].astype(np.int64)

toss_winner_advantage = 0

# This checks only the first 20 matches
for i, mid in enumerate(match_ids_matches[:20]):
    toss_winner = toss_winners[i]

    toss_runs = np.sum(batsman_runs[(match_ids_del == mid) & (batting_team == toss_winner)])
    opp_runs = np.sum(batsman_runs[(match_ids_del == mid) & (batting_team != toss_winner)])

    if toss_runs > opp_runs:
        toss_winner_advantage += 1

print("Toss Impact Analysis (First 20 Matches)")
print("-" * 50)
print(f"Matches won by toss winner: {toss_winner_advantage}/20")
print(f"Toss winner advantage: {(toss_winner_advantage / 20) * 100:.1f}%")
```

## Task 11: Match Scorecard Generation

```python
match_id = deliveries_np[1:, 0].astype(np.int64)
batting_team = deliveries_np[1:, 2]
batsman_runs = deliveries_np[1:, 9].astype(np.int64)

unique_matches = np.unique(match_id)

print("Match Scorecard (First 10 Matches)")
print("=" * 50)

for mid in unique_matches[:10]:
    teams = np.unique(batting_team[match_id == mid])

    print(f"\nMatch {mid}:")
    print("-" * 50)

    team_scores = []

    for team in teams:
        runs = np.sum(batsman_runs[(match_id == mid) & (batting_team == team)])
        team_scores.append((team, runs))
        print(f"  {team}: {runs} runs")

    if len(team_scores) == 2:
        if team_scores[0][1] > team_scores[1][1]:
            print(f"  Winner: {team_scores[0][0]}")
        elif team_scores[1][1] > team_scores[0][1]:
            print(f"  Winner: {team_scores[1][0]}")
        else:
            print("  Result: Tie")
```
