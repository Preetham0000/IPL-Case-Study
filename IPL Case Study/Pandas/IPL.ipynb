## Stage 1: Data Ingestion

```python
# Import required libraries
import pandas as pd
import numpy as np

# Define file paths
deliveries_path = '../Dataset/deliveries.csv'
matches_path = '../Dataset/matches.csv'

# Load datasets into DataFrames
deliveries_df = pd.read_csv(deliveries_path)
matches_df = pd.read_csv(matches_path)

print("Datasets loaded successfully!")
```

### Inspect Deliveries DataFrame

```python
# Display shape
print(f"Shape: {deliveries_df.shape}")
print(f"Rows: {deliveries_df.shape[0]}, Columns: {deliveries_df.shape[1]}\n")

# Display columns
print("Columns:")
print(deliveries_df.columns.tolist())
print()

# Display data types
print("Data Types:")
print(deliveries_df.dtypes)
```

### Inspect Matches DataFrame

```python
# Display shape
print(f"Shape: {matches_df.shape}")
print(f"Rows: {matches_df.shape[0]}, Columns: {matches_df.shape[1]}\n")

# Display columns
print("Columns:")
print(matches_df.columns.tolist())
print()

# Display data types
print("Data Types:")
print(matches_df.dtypes)
```
